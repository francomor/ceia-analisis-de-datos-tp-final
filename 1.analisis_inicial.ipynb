# Hallazgos del Dataset de Crímenes de Chicago 2025

```python
# Contar la proporción de valores nulos en cada columna
df.isna().sum()  
```

```python
# Registros con valores nulos en coordenadas/ubicación
registros_sin_ubicacion = df[df['X Coordinate'].isna() | df['Y Coordinate'].isna() | 
                             df['Latitude'].isna() | df['Longitude'].isna() | 
                             df['Location'].isna()]
print(f"Total de registros con valores nulos en coordenadas: {len(registros_sin_ubicacion)}")
registros_sin_ubicacion
```

```python
# 1. Features con más del 40% de valores nulos
total_rows = len(df)
null_percentages = (df.isna().sum() / total_rows * 100).round(2)
features_high_null = null_percentages[null_percentages > 40].sort_values(ascending=False)
print(f"Features con más del 40% de valores nulos ({len(features_high_null)} features):")
print(features_high_null.to_string())
```

```python
# 2. Distribución de Arrestos (desbalance)
arrest_counts = df['Arrest'].value_counts()
arrest_pct = df['Arrest'].value_counts(normalize=True) * 100
print("Distribución de Arrestos:")
for val, count in arrest_counts.items():
    print(f"  {val}: {count:,} ({arrest_pct[val]:.1f}%)")
```

```python
# 2b. Distribución de Crímenes Domésticos (desbalance)
domestic_counts = df['Domestic'].value_counts()
domestic_pct = df['Domestic'].value_counts(normalize=True) * 100
print("Distribución de Crímenes Domésticos:")
for val, count in domestic_counts.items():
    print(f"  {val}: {count:,} ({domestic_pct[val]:.1f}%)")
```

```python
# Relación entre Arrestos y Crímenes Domésticos
cross_tab = pd.crosstab(df['Domestic'], df['Arrest'], normalize='index') * 100

fig, ax = plt.subplots(figsize=(10, 6))
x = range(len(cross_tab.index))
width = 0.35

bars1 = ax.bar([i - width/2 for i in x], cross_tab[False], width, label='Sin Arresto', color='steelblue')
bars2 = ax.bar([i + width/2 for i in x], cross_tab[True], width, label='Con Arresto', color='coral')

ax.set_xlabel('Crimen Doméstico')
ax.set_ylabel('Porcentaje (%)')
ax.set_title('Relación entre Arrestos y Crímenes Domésticos')
ax.set_xticks(x)
ax.set_xticklabels(['No Doméstico', 'Doméstico'])
ax.legend()

# Agregar etiquetas de porcentaje
for bar in bars1:
    height = bar.get_height()
    ax.annotate(f'{height:.1f}%', xy=(bar.get_x() + bar.get_width()/2, height),
                xytext=(0, 3), textcoords="offset points", ha='center', va='bottom')
for bar in bars2:
    height = bar.get_height()
    ax.annotate(f'{height:.1f}%', xy=(bar.get_x() + bar.get_width()/2, height),
                xytext=(0, 3), textcoords="offset points", ha='center', va='bottom')

plt.tight_layout()
plt.show()
```

```python
# Top 10 tipos de crímenes con mayor tasa de arrestos (por código IUCR)
# Calcular tasa de arresto por IUCR (solo IUCR con al menos 50 casos para significancia)
iucr_total = df.groupby('IUCR').size()
iucr_arrestos = df[df['Arrest'] == True].groupby('IUCR').size()
tasa_arresto = (iucr_arrestos / iucr_total * 100).dropna()

# Filtrar solo IUCR con al menos 50 casos
iucr_significativos = iucr_total[iucr_total >= 50].index
tasa_arresto_filtrada = tasa_arresto[tasa_arresto.index.isin(iucr_significativos)]
top_tasa_arresto = tasa_arresto_filtrada.sort_values(ascending=False).head(30)

# Crear etiquetas con Primary Type - Description
etiquetas_arrest = []
for iucr in top_tasa_arresto.index:
    info = df[df['IUCR'] == iucr][['Primary Type', 'Description']].iloc[0]
    etiquetas_arrest.append(f"{info['Primary Type']} - {info['Description']}")

plt.figure(figsize=(14, 6))
bars = plt.bar(range(len(top_tasa_arresto)), top_tasa_arresto.values, color='coral')
plt.xticks(range(len(top_tasa_arresto)), etiquetas_arrest, rotation=45, ha='right')
plt.title('Top 10 Tipos de Crímenes con Mayor Tasa de Arrestos (%)')
plt.xlabel('Tipo de Crimen - Descripción')
plt.ylabel('Tasa de Arrestos (%)')

# Agregar etiquetas de porcentaje
for bar in bars:
    height = bar.get_height()
    plt.annotate(f'{height:.1f}%', xy=(bar.get_x() + bar.get_width()/2, height),
                xytext=(0, 3), textcoords="offset points", ha='center', va='bottom')

plt.tight_layout()
plt.show()
```

```python
# 3. Top Community Areas con mayor cantidad de crímenes
top_areas = df['Community Area'].value_counts().head(10)
print("Top 10 Community Areas con mayor cantidad de crímenes:")
for area, count in top_areas.items():
    print(f"  Community Area {int(area)}: {count:,} crímenes")
```

```python
# 4. Top 10 tipos de crímenes más frecuentes (Primary Type)
top_crimes = df['Primary Type'].value_counts().head(10)
print("Top 10 tipos de crímenes más frecuentes:")
for crime, count in top_crimes.items():
    print(f"  {crime}: {count:,}")
```

```python
# 5. Resumen de hallazgos
print("=" * 60)
print("RESUMEN DE HALLAZGOS - Dataset Crímenes Chicago 2025")
print("=" * 60)

# Total de registros
print(f"\n📊 Total de registros: {len(df):,}")

# Valores nulos
null_pct = (df.isna().sum() / len(df) * 100)
high_null = null_pct[null_pct > 40]
print(f"\n🔴 Features con >40% valores nulos: {len(high_null)}")
if len(high_null) > 0:
    for col in high_null.index:
        print(f"   - {col}: {high_null[col]:.1f}%")

# Desbalance en Arrests
arrest_pct = df['Arrest'].value_counts(normalize=True) * 100
print(f"\n⚖️ Desbalance en Arrestos:")
print(f"   - Sin arresto: {arrest_pct.get(False, 0):.1f}%")
print(f"   - Con arresto: {arrest_pct.get(True, 0):.1f}%")

# Top crímenes
print(f"\n🔝 Top 5 crímenes más frecuentes:")
for crime, count in df['Primary Type'].value_counts().head(5).items():
    print(f"   - {crime}: {count:,}")

# Top Community Areas
print(f"\n📍 Top 5 Community Areas con más crímenes:")
for area, count in df['Community Area'].value_counts().head(5).items():
    print(f"   - Área {int(area)}: {count:,}")
```

```python
# Histograma de casos por mes
df['Date'] = pd.to_datetime(df['Date'])
df['Month'] = df['Date'].dt.month

casos_por_mes = df['Month'].value_counts().sort_index()

plt.figure(figsize=(10, 6))
plt.bar(casos_por_mes.index, casos_por_mes.values, color='steelblue')
plt.title('Cantidad de Crímenes por Mes')
plt.xlabel('Mes')
plt.ylabel('Cantidad de Crímenes')
plt.xticks(range(1, 13), ['Ene', 'Feb', 'Mar', 'Abr', 'May', 'Jun', 'Jul', 'Ago', 'Sep', 'Oct', 'Nov', 'Dic'])
plt.tight_layout()
plt.show()
```
